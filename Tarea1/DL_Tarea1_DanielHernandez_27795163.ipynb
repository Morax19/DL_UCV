## **Parte II: Evaluación**

### **II-A.** Backpropagation:

Sabemos que este algoritmo puede dividirse en 2 pasos, el **forward pass** (paso hacia adelante) y el **backward pass** (paso hacia atrás), por tanto tenemos que:



* **Forward Pass:**
Reescribimos la red neuronal como serie de cálculos secuenciales, para ello vamos a definir el vector de entrada:
<br><br>
$X = \begin{bmatrix} x_1 \\ x_2 \end{bmatrix}$
<br><br>
Para cada una de las capas presentes, tenemos los siguientes calculos:
<br><br>
$f_0 = β_0 + Ω_0 X$
<br><br>
$h_1 = σ[f_0]$
<br><br>
$f_1 = β_1 + Ω_1 h_1$
<br><br>
$h_2 = σ[f_1]$
<br><br>
$f_2 = β_2 + Ω_2 h_2$
<br><br>
Finalmente, sabiendo que nuestra función objetivo es el $MSE$, calculamos la pérdida de la siguiente manera:
<br><br>
$ℓ_i = 𝐿[f_2, y_i]$
<br><br>
$ℓ_i = \frac{1}{n} ∑{(f_2 - y_i)^2}$

* **Backward Pass**:
Debemos calcular las derivadas parciales de la función objetivo con respecto a la salida obtenida por el modelo y recorrer las capas en reversa, sabiendo que la derivada de nuestra función de activación es:
<br><br>
$\frac{∂σ(t)}{∂t} = σ(t)(1 - σ(t))$
<br><br>
Por lo tanto:
<br><br>
$
\frac{∂ℓ_i}{∂f_2} = \frac{2(f_2 - y)}{n}
$
<br><br>
$
\frac{∂ℓ_i}{∂h_2} = \frac{∂f_2}{∂h_2}\frac{∂ℓ_i}{∂f_2}
⟹ \frac{∂(β_2 + Ω_2h_2)}{∂ h_2}\frac{∂ℓ_i}{∂f_2} = Ω_2^T\frac{∂ℓ_i}{∂f_2}
$
<br><br>
$
\frac{∂ℓ_i}{∂f_1} = \frac{∂h_2}{∂f_1}(\frac{∂f_2}{∂h_2}\frac{∂ℓ_i}{∂f_2})
⟹ \frac{∂σ[f_1]}{∂f_1}(\frac{∂f_2}{∂h_2}\frac{∂ℓ_i}{∂f_2})
= σ(f_1)(1 - σ(f_1))(\frac{∂f_2}{∂h_2}\frac{∂ℓ_i}{∂f_2})
$
<br><br>
$
\frac{∂ℓ_i}{∂h_1} = \frac{∂f_1}{∂h_1}(\frac{∂h_2}{∂f_1}\frac{∂f_2}{∂h_2}\frac{∂ℓ_i}{∂f_2})
⟹ \frac{∂(β_1 + Ω_1h_1)}{∂ h_1}(\frac{∂h_2}{∂f_1}\frac{∂f_2}{∂h_2}\frac{∂ℓ_i}{∂f_2})
=Ω_1^T(\frac{∂h_2}{∂f_1}\frac{∂f_2}{∂h_2}\frac{∂ℓ_i}{∂f_2})
$
<br><br>
$\frac{∂ℓ_i}{∂f_0}
=\frac{∂h_1}{∂f_0}(\frac{∂f_1}{∂h_1}\frac{∂h_2}{∂f_1}\frac{∂f_2}{∂h_2}\frac{∂ℓ_i}{∂f_2})
⟹ \frac{∂σ[f_0]}{∂f_0}(\frac{∂f_1}{∂h_1}\frac{∂h_2}{∂f_1}\frac{∂f_2}{∂h_2}\frac{∂ℓ_i}{∂f_2})
= σ(f_0)(1 - σ(f_0))(\frac{∂f_1}{∂h_1}\frac{∂h_2}{∂f_1}\frac{∂f_2}{∂h_2}\frac{∂ℓ_i}{∂f_2})
$
<br><br>
De la misma manera, calculamos las derivadas parciales con respecto a $β_k$ y a $Ω_k$:
<br><br>
$
\frac{∂ℓ_i}{∂β_2} = \frac{∂f_2}{∂β_2}\frac{∂ℓ_i}{∂f_2}
⟹ \frac{∂(β_2 + Ω_2h_2)}{∂ \beta_2}\frac{∂ℓ_i}{∂f_2} = \frac{∂ℓ_i}{∂f_2}
$
<br><br>
$
\frac{∂ℓ_i}{∂β_1} = \frac{∂f_1}{∂β_1}\frac{∂ℓ_i}{∂f_1}
⟹ \frac{∂(β_1 + Ω_1h_1)}{∂ \beta_1}\frac{∂ℓ_i}{∂f_1} = \frac{∂ℓ_i}{∂f_1}
$
<br><br>
$
\frac{∂ℓ_i}{∂β_0} = \frac{∂f_0}{∂β_0}\frac{∂ℓ_i}{∂f_0}
⟹ \frac{∂(β_0 + Ω_0h_0)}{∂ \beta_0}\frac{∂ℓ_i}{∂f_0} = \frac{∂ℓ_i}{∂f_0}
$
<br><br>
$
\frac{∂ℓ_i}{∂Ω_2} = \frac{∂f_2}{∂Ω_2}\frac{∂ℓ_i}{∂f_2}
⟹ \frac{∂(β_2 + Ω_2h_2)}{∂Ω_2}\frac{∂ℓ_i}{∂f_2} = \frac{∂ℓ_i}{∂f_2}(h_2^T)
$
<br><br>
$
\frac{∂ℓ_i}{∂Ω_1} = \frac{∂f_1}{∂Ω_1}\frac{∂ℓ_i}{∂f_1}
⟹ \frac{∂(β_1 + Ω_1h_1)}{∂Ω_1}\frac{∂ℓ_i}{∂f_1} = \frac{∂ℓ_i}{∂f_1}(h_1^T)
$
<br><br>
$
\frac{∂ℓ_i}{∂Ω_0} = \frac{∂f_0}{∂Ω_0}\frac{∂ℓ_i}{∂f_0}
⟹ \frac{∂(β_0 + Ω_0X)}{∂Ω_0}\frac{∂ℓ_i}{∂f_0} = \frac{∂ℓ_i}{∂f_0}(X^T)
$

De forma general (y modificando la ecuación 7.25 del libro), nos quedarían las siguientes expresiones:
<br><br>
$\frac{∂ℓ_i}{∂β_k} = \frac{∂ℓ_i}{∂f_k} \qquad k ∈ \{K, K-1, …, 1\}$
<br><br>
$\frac{∂ℓ_i}{∂Ω_k} = \frac{∂ℓ_i}{∂f_k}h_k^T \qquad k ∈ \{K, K-1, …, 1\}$
<br><br>
$\frac{∂ℓ_i}{∂f_{k-1}} = σ(f_{k-1})(1 - σ(f_{k-1}))⊙(Ω_k^T\frac{∂ℓ_i}{∂f_k}) \qquad k ∈ \{K, K-1, …, 1\}$
<br><br>
